```python
df["Price"] = df["Price"].str.replace("₹", "", regex=False)
df["Price"] = df["Price"].str.replace(",", "", regex=False)
df["Price"] = df["Price"].astype(int)
```

```python
df.loc[
    df["Ram"].str.contains("Storage:|SSD|Hard Disk", case=False, na=False),
    ["Model", "Ram", "SSD", "Generation", "Core"]
]
```

```python
bad_ram = df["Ram"].str.contains(
    "Storage:|SSD|Hard Disk",
    case=False,
    na=False
)
```

```python
df.loc[bad_ram, ["Ram", "Core"]]
```

```python
df.loc[bad_ram & df["Core"].str.contains("GB", na=False), "Ram"] = \
    df.loc[bad_ram & df["Core"].str.contains("GB", na=False), "Core"]
```

```python
df.loc[679, "Ram"] = None
```

```python
df.loc[bad_ram, ["Ram", "Core"]]
```

```python
df["Ram_GB"] = df["Ram"].str.extract(r"(\d+)\s*GB", expand=False)
df["Ram_GB"] = pd.to_numeric(df["Ram_GB"], errors="coerce")
```

```python
df["Ram_Type"] = df["Ram"].str.extract(
    r"(LPDDR\dX?|DDR\dX?|SDRAM)",
    expand=False
)
```

```python
df.to_csv("../data/processed/laptop_cleaned_v1.csv", index=False)
```
